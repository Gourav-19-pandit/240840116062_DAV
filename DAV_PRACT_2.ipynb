```python
# importing  pandas as pd
import pandas as pd
```

```python
# created the data set for creation of  dataframe
data={

       "ID": [ 101, 102, 103, 104,104, 105, 106, 107],
       "Name" :["Rahul", "Priya", "Amit", None, "Riya","Ajay", "Naresh","Jenil"],
       "Age" : [ 20, 21,None , 20,20 , 22, 23, 28],
       "City" :["Surat", "Navsari", "Surat", "Bharuch", "Bharuch","Navsari", "Surat","Ahmedabad"],
       "Marks" :[ 85,None , 78,88 ,88, 81, 90,93]
}
df = pd.DataFrame(data)
print(df)
```

```python
# check missing values in the datset
df.isnull()
```

```python
# print the count of null values for each columns
df.isnull().sum()
```

```python
# print the total no of null value in whole tabel
df.isnull().sum().sum()
```

```python
# fill the missing values in tabel in marks coloumn
df["Marks"] = df["Marks"].fillna(df["Marks"].mean())
df
```

```python
#replace missing age values  in the age column
df["Age"] = df["Age"].fillna(df["Age"].mean())
df
```

```python
# replace the null values in name column
df["Name"] = df["Name"].fillna("Gourav")
df
```

```python
#detect duplicate records
df.duplicated()
```

```python
# count the total no of duplicated rows
df.duplicated().sum()
```

```python
#remove te duplicated records
df.drop_duplicates()
```

```python
#rename the coloumn marks to score
df.rename(columns={"Marks":"Score"}, inplace=True)
df
```

```python
 # changing the data type age coloumn float to integer

df['Age'] = df['Age'].astype(int)

display(df)
```

```python
 # create a new column on final score  in existing data frame
 df["Final_Score"]=df["Score"]+5
 df
```

```python
# convert the students coloumn in uppper case
df["Name"]=df["Name"].str.upper()
display(df)
```

```python
# convert the City coloumn in lower case
df["City"]=df["City"].str.lower()
df
```

```python
# remove extra space from the tabel column "city"
df["City"]=df["City"].str.strip()
display(df)
```

```python
## remove leading  extra space from the tabel column "city"
df["City"]=df["City"].str.rstrip()
display(df)
```

```python
## remove traling  extra space from the tabel column "city"
df["City"]=df["City"].str.lstrip()
display(df)
```

```python
#sort the dataframe based on the final score column
df.sort_values(by="Final_Score", ascending=False)
```

```python
from numpy import False_
# save the clean dataset
df.to_csv("cleaned_Students.csv", index=False_)
df
```
